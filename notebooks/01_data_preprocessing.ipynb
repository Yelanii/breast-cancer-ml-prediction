```python
df.drop(columns=['Occupation','Patient_ID','Month_of_Birth'], inplace=True)
```

```python
df = df.dropna()
df.isnull().sum()
```

```python
unique_counts = df.nunique()
print(unique_counts)
```

```python
df['Sex'].unique()
```

```python
df['T_Stage'] = df['T_Stage'].map({'T1': 1, 'T2': 2, 'T3': 3, 'T4': 4})
df['N_Stage'] = df['N_Stage'].map({'N1': 1, 'N2': 2, 'N3': 3})
df['Estrogen_Status'] = df['Estrogen_Status'].map({'Positive': 1, 'Negative': 0})
df['Progesterone_Status'] = df['Progesterone_Status'].map({'Positive': 1, 'Negative': 0})
df['A_Stage'] = df['A_Stage'].map({'Regional': 0, 'Distant': 1})
df['Differentiated'] = df['Differentiated'].map({'Well differentiated': 0, 'Moderately differentiated': 1, 'Poorly differentiated': 2, 'Undifferentiated': 3})
df['6th_Stage'] = df['6th_Stage'].map({'IIA': 1, 'IIB': 2, 'IIIA': 3, 'IIIB': 4, 'IIIC': 5})

df.head()
```

```python
df.drop(columns=['Sex'], inplace=True)
```

```python
df.head()
```

```python
df['Mortality_Status'].unique()
```

```python
df['Mortality_Status'] = df['Mortality_Status'].map({'Alive': 1, 'Dead': 0, 'ALIVE': 1, 'DEAD': 0, 'ALive':1, 'alive':1, 'dead':0})
```

```python
for col in df.columns:
    if df[col].nunique() < 10:
        print(f"\nColumn: {col}")
        value_counts = df[col].value_counts(dropna=False)
        percent = (value_counts / len(df)) * 100
        summary = pd.DataFrame({'Count': value_counts, 'Percentage': percent.round(2)})
        print(summary)
```

```python
import matplotlib.pyplot as plt

df['Mortality_Status'].value_counts().plot(kind='bar', color=['green', 'red'])
plt.title('Distribution of Mortality Status')
plt.xlabel('Status (0 = Alive, 1 = Dead)')
plt.ylabel('Number of Patients')
plt.show()

```

```python
Survival_Months = df['Survival_Months']
Morality_Status = df['Mortality_Status']
df.drop(columns=['Survival_Months','Mortality_Status'], inplace=True)
```

```python
df_with_survival = df.copy()
df_with_survival['Survival_Months'] = Survival_Months

df_with_mortality = df.copy()
df_with_mortality['Mortality_Status'] = Morality_Status

# Save to CSV files
df_with_survival.to_csv('regression_data.csv', index=False)
df_with_mortality.to_csv('classification_data.csv', index=False)
```
